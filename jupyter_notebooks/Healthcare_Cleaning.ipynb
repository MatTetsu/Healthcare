```python
df = df.iloc[:10000].copy()
to_drop = []
for idx, age in df["Age"].items():
    if age < 18:
        to_drop.append(idx)
    elif 18 <= age < 30:
        df.loc[idx, "Age Range"] = "18-29"
    elif 30 <= age < 45:
        df.loc[idx, "Age Range"] = "30-44"
    elif 45 <= age < 60:
        df.loc[idx, "Age Range"] = "45-59"
    else:
        df.loc[idx, "Age Range"] = "60+"
df.drop(index=to_drop, inplace=True)
```

Here I have made an if statement to drop each person under the age of 18, and add age ranges for everyone over 18. The reason I have dropped everyone who was under the age of 18 was for ethical reasons.

```python
def create_download_link(df, title="healthcare_cleaned.csv", filename="healthcare_cleaned.csv"):
    """Generates a link to download the data in a given pandas DataFrame"""
    csv = df.to_csv()
    b64 = base64.b64encode(csv.encode())
    payload = b64.decode()
    html = '<a download="{filename}" href="data:text/csv;base64,{payload}" target="_blank">{title}</a>'
    html = html.format(payload=payload, title=title, filename=filename)
    return HTML(html)
create_download_link(df)
```

```python
def create_download_link_xlsx(df, title="healthcare_cleaned.xlsx", filename="healthcare_cleaned.xlsx"):
    """Generates a link to download the data in a given pandas DataFrame as XLSX"""
    towrite = io.BytesIO()
    df.to_excel(towrite, index=False)
    towrite.seek(0)
    b64 = base64.b64encode(towrite.read()).decode()
    html = f'<a download="{filename}" href="data:application/vnd.openxmlformats-officedocument.spreadsheetml.sheet;base64,{b64}" target="_blank">{title}</a>'
    return HTML(html)
create_download_link_xlsx(df)
```
